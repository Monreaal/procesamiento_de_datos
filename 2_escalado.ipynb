# Escalado
Conjunto de datos: iris (incluido en sklearn)
Este conjunto de datos es compacto y perfecto para demostrar técnicas de escalado.

```python
# Importar librerías necesarias
from sklearn.datasets import load_iris
from sklearn.preprocessing import MinMaxScaler, StandardScaler
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Cargar el conjunto de datos
iris = load_iris(as_frame=True)
data = iris.data
```

```python
# Visualizar las estadísticas iniciales
print("Estadísticas iniciales:")
data.describe()
```

```python
# Aplicar MinMaxScaler
minmax_scaler = MinMaxScaler()
data_minmax = pd.DataFrame(minmax_scaler.fit_transform(data), columns=data.columns)
```

```python
# Aplicar StandardScaler
standard_scaler = StandardScaler()
data_standard = pd.DataFrame(standard_scaler.fit_transform(data), columns=data.columns)
```

### 1. ¿Qué diferencias encuentras entre las estadísticas originales y las del escalado MinMax?


Logramos observar que el valor minimo se convirtio en "0" mientras el maximo en "1" y en base a estos se reajustaron el resto de valores 

```python
# Estadísticas de los datos originales
print("--- ANTES DEL ESCALADO ---")
display(data[[ 'sepal length (cm)', 'sepal width (cm)', 'petal length (cm)', 'petal width (cm)' ]].describe().round(2))

# Estadísticas de los datos escalados
print("\n--- DESPUÉS DEL ESCALADO ---")
display(data_minmax[[ 'sepal length (cm)', 'sepal width (cm)', 'petal length (cm)', 'petal width (cm)' ]].describe().round(2))
```

### 2. Crea una gráfica de la distribución de la variable 'sepal length (cm)' antes y después del escalado.

```python
plt.figure(figsize=(10, 6))


sns.histplot(data=data, x='sepal length (cm)', kde=True, color='blue', alpha=0.5, label='Sepal Length Original')
# x = 'sepal length (cm)' es el nombre de la columna que queremos graficar
# kde=True agrega una curva de densidad al histograma
# alpha=0.5 hace que el histograma sea semi-transparente

sns.histplot(data=data_minmax, x='sepal length (cm)', kde=True, color='red', alpha=0.5, label='Sepal Length Escalado')


plt.title('Comparación de Distribución de Sepal Length')
plt.xlabel('Longitud del Sépalo (cm)')
plt.ylabel('Frecuencia')
plt.legend()

plt.show()
```

### 3. Crea una gráfica de dispersión entre 'sepal length (cm)' y 'sepal width (cm)' con los datos escalados por StandardScaler.

```python
plt.figure(figsize=(10, 6))


sns.histplot(data=data, x='sepal length (cm)', kde=True, color='blue', alpha=0.5, label='Sepal Length Original')

sns.histplot(data=data_standard, x='sepal length (cm)', kde=True, color='red', alpha=0.5, label='Sepal Length Escalado')


plt.title('Comparación de Distribución de Sepal Length')
plt.xlabel('Longitud del Sépalo (cm)')
plt.ylabel('Frecuencia')
plt.legend()

plt.show()
```
